**Набор данных:** [Students Performance in Exams](https://www.kaggle.com/datasets/spscientist/students-performance-in-exams) (Kaggle).

Данные содержат результаты экзаменов 1000 школьников по трём предметам и их характеристики:

| Признак | Тип | Описание |
|---|---|---|
| `gender` | категориальный | пол (female / male) |
| `race/ethnicity` | категориальный | этническая группа (A–E) |
| `parental level of education` | категориальный | образование родителей |
| `lunch` | категориальный | тип обеда (standard / free/reduced — косвенный признак дохода семьи) |

| `math score`, `reading score`, `writing score` | непрерывные (0–100) | баллы за экзамены |

**Почему этот набор подходит:** есть непрерывный признак (`math score`) для доверительных интервалов,
естественные группы для двухвыборочного теста (пол, подготовительный курс),
**парные наблюдения** (у каждого ученика есть и балл за чтение, и балл за письмо) для парного t-теста,
и несколько категориальных признаков для критерия хи-квадрат на независимость.

**Основной исследуемый признак:** `math score` — балл за экзамен по математике.

```python
print('Пропуски по столбцам:')
print(df.isna().sum())
print(f'\nПолных дубликатов строк: {df.duplicated().sum()}')

score_cols = ['math', 'reading', 'writing']
cat_cols = ['gender', 'race', 'parent_edu', 'lunch', 'prep']

# Баллы должны лежать в диапазоне [0, 100]
out_of_range = ((df[score_cols] < 0) | (df[score_cols] > 100)).sum()
print('\nЗначений вне диапазона [0, 100]:')
print(out_of_range)
```

## 3. Параметрические доверительные интервалы

**Для среднего** (дисперсия неизвестна) — t-интервал:
$\bar{x} \pm t_{1-\alpha/2,\,n-1}\cdot\dfrac{s}{\sqrt n}$.

**Для дисперсии** — интервал на основе $\chi^2$-распределения:
$\left(\dfrac{(n-1)s^2}{\chi^2_{1-\alpha/2,\,n-1}},\ \dfrac{(n-1)s^2}{\chi^2_{\alpha/2,\,n-1}}\right)$.
Этот интервал **требует нормальности данных** и к ней гораздо чувствительнее, чем t-интервал.

```python
def t_ci(data, alpha=0.05):
    # Параметрический ДИ для среднего на основе t-распределения
    data = np.asarray(data, dtype=float)
    n = len(data)
    m, se = data.mean(), stats.sem(data)
    t_crit = stats.t.ppf(1 - alpha / 2, df=n - 1)
    return m - t_crit * se, m + t_crit * se

def var_ci(data, alpha=0.05):
    # Параметрический ДИ для дисперсии на основе chi2-распределения
    data = np.asarray(data, dtype=float)
    n, s2 = len(data), np.var(data, ddof=1)
    return ((n - 1) * s2 / stats.chi2.ppf(1 - alpha / 2, n - 1),
            (n - 1) * s2 / stats.chi2.ppf(alpha / 2, n - 1))

ci95 = t_ci(x, 0.05)
ci99 = t_ci(x, 0.01)
vci95 = var_ci(x, 0.05)

print(f'Выборочное среднее: {x.mean():.3f}, SE = {stats.sem(x):.3f}')
print(f'95% t-ДИ для среднего: ({ci95[0]:.3f}, {ci95[1]:.3f}), ширина = {ci95[1]-ci95[0]:.3f}')
print(f'99% t-ДИ для среднего: ({ci99[0]:.3f}, {ci99[1]:.3f}), ширина = {ci99[1]-ci99[0]:.3f}')
print(f'\nВыборочная дисперсия: {x.var(ddof=1):.3f} (s = {x.std(ddof=1):.3f})')
print(f'95% chi2-ДИ для дисперсии: ({vci95[0]:.3f}, {vci95[1]:.3f})')
print(f'95% ДИ для ст. отклонения: ({np.sqrt(vci95[0]):.3f}, {np.sqrt(vci95[1]):.3f})')

# Сверка с готовой функцией scipy
print('\nПроверка через scipy:', stats.t.interval(0.95, n - 1, loc=x.mean(), scale=stats.sem(x)))
```

**Интерпретация.** С доверительной вероятностью 95% средний балл по математике в генеральной совокупности лежит в интервале **(65.15; 67.03)**, с вероятностью 99% — в **(64.85; 67.33)**. 99%-интервал шире (2.48 против 1.88 балла): за большую уверенность платим точностью. Оба интервала узкие благодаря большому объёму выборки (SE = 0.48).

Интервал для стандартного отклонения: (14.53; 15.86). Его следует воспринимать осторожно: χ²-интервал выведен в предположении нормальности, а данные имеют тяжёлый левый хвост, поэтому реальная вероятность покрытия может отличаться от 95%.

## 4. Бутстреп-доверительные интервалы

Улучшения по сравнению с исходным шаблоном:

* бутстреп **векторизован**: все $B$ псевдовыборок генерируются одной матрицей индексов — в десятки раз быстрее цикла;
* нормальный интервал центрируется на оценке по **исходной** выборке $\hat\theta$ (как в формуле $\hat\theta \pm z_{1-\alpha/2}\cdot SE_{boot}$), а не на среднем бутстреп-оценок;
* дополнительно считается оценка **смещения** и **BCa-интервал** (через `scipy.stats.bootstrap`).

```python
def bootstrap_ci(data, statistic, n_bootstrap=B, ci=95, rng=rng):
    # Бутстреп: возвращает нормальный и процентильный ДИ, а также сами бутстреп-оценки.
    # statistic должна принимать параметр axis (np.mean, np.median и т.п.).
    data = np.asarray(data, dtype=float)
    n = len(data)
    idx = rng.integers(0, n, size=(n_bootstrap, n))     # B псевдовыборок с возвращением
    estimates = statistic(data[idx], axis=1)

    theta_hat = statistic(data)                         # оценка по исходной выборке
    se_boot = estimates.std(ddof=1)
    alpha = 1 - ci / 100
    z = stats.norm.ppf(1 - alpha / 2)

    normal_ci = (theta_hat - z * se_boot, theta_hat + z * se_boot)
    perc_ci = tuple(np.percentile(estimates, [100 * alpha / 2, 100 * (1 - alpha / 2)]))
    bias = estimates.mean() - theta_hat
    return {'theta': theta_hat, 'se': se_boot, 'bias': bias,
            'normal': normal_ci, 'percentile': perc_ci, 'estimates': estimates}

def bca_ci(data, statistic, n_bootstrap=B, ci=95):
    # BCa-интервал (scipy сам считает z0 и ускорение a через jackknife)
    res = stats.bootstrap((np.asarray(data, dtype=float),), statistic,
                          n_resamples=n_bootstrap, confidence_level=ci / 100,
                          method='BCa', random_state=42)
    return (res.confidence_interval.low, res.confidence_interval.high)

boot_mean = bootstrap_ci(x, np.mean)
boot_median = bootstrap_ci(x, np.median)
bca_mean = bca_ci(x, np.mean)
# BCa для медианы не считаем: баллы целые, и jackknife-медианы (выборки без одного
# наблюдения) почти все совпадают -> параметр ускорения a = 0/0 и интервал вырождается в NaN.

for name, r in [('Среднее', boot_mean), ('Медиана', boot_median)]:
    print(f'=== {name}: оценка = {r["theta"]:.3f}, SE_boot = {r["se"]:.3f}, смещение = {r["bias"]:+.4f}')
    print(f'  Нормальный 95%:    ({r["normal"][0]:.3f}, {r["normal"][1]:.3f})')
    print(f'  Процентильный 95%: ({r["percentile"][0]:.3f}, {r["percentile"][1]:.3f})')
print(f'BCa 95% для среднего: ({bca_mean[0]:.3f}, {bca_mean[1]:.3f})')
print(f'Уникальных значений бутстреп-медианы: {np.unique(boot_median["estimates"]).size}')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Бутстреп-распределение среднего
ax = axes[0]
est = boot_mean['estimates']
ax.hist(est, bins=50, density=True, alpha=0.6, color='skyblue', edgecolor='white')
g = np.linspace(est.min(), est.max(), 300)
ax.plot(g, stats.norm.pdf(g, boot_mean['theta'], boot_mean['se']), 'k-', lw=1.5, label='Нормальная аппроксимация')
ax.axvline(boot_mean['theta'], color='black', ls=':', label=f'Исходное среднее ({boot_mean["theta"]:.2f})')
for v in boot_mean['percentile']:
    ax.axvline(v, color='red', ls='--')
for v in ci95:
    ax.axvline(v, color='green', ls='-.')
ax.plot([], [], 'r--', label='Процентильный 95% ДИ')
ax.plot([], [], 'g-.', label='Параметрический 95% t-ДИ')
ax.set_title(f'Бутстреп-распределение среднего (B = {B})')
ax.set_xlabel('Среднее math score'); ax.legend(fontsize=9)

# Бутстреп-распределение медианы
ax = axes[1]
est = boot_median['estimates']
vals, counts = np.unique(est, return_counts=True)
ax.bar(vals, counts / counts.sum(), width=0.4, color='plum')
for v in boot_median['percentile']:
    ax.axvline(v, color='red', ls='--')
ax.axvline(boot_median['theta'], color='black', ls=':', label=f'Исходная медиана ({boot_median["theta"]:.1f})')
ax.plot([], [], 'r--', label='Процентильный 95% ДИ')
ax.set_title('Бутстреп-распределение медианы (дискретное!)')
ax.set_xlabel('Медиана math score'); ax.set_ylabel('Доля'); ax.legend(fontsize=9)

plt.tight_layout()
plt.show()
```

**Сравнение и объяснение различий.**

Гистограмма бутстреп-средних симметрична и хорошо описывается нормальной кривой. Причина — большой объём выборки и умеренная асимметрия: по ЦПТ распределение выборочного среднего уже нормально, и параметрический метод не проигрывает бутстрепу. BCa чуть сдвинут влево — он учитывает слабую левую асимметрию данных.
* **Медиана.** Процентильный интервал **[65; 67]**. Он немного шире интервала для среднего, т.к. при близком к нормальному распределению медиана менее эффективна (SE_boot = 0.70 против 0.48). Бутстреп-распределение медианы **дискретно** (баллы целые, медиана принимает лишь значения вида k и k+0.5), поэтому нормальный бутстреп-интервал для неё некорректен, а BCa вообще не вычисляется (jackknife-оценки медианы одинаковы, параметр ускорения не определён). Для медианы правильный выбор — процентильный метод.
* Параметрической формулы для ДИ медианы «из коробки» нет — это как раз ситуация, где бутстреп незаменим.

## 5. Проверка гипотез

Во всех тестах уровень значимости $\alpha = 0.05$. Кроме p-значения приводим **доверительный интервал**
для эффекта и **размер эффекта** (Cohen's d, V Крамера), а в конце делаем поправку на множественные сравнения.

```python
def cohen_d_ind(a, b):
    # Cohen's d для независимых выборок (по формуле из методички)
    return (np.mean(a) - np.mean(b)) / np.sqrt((np.var(a, ddof=1) + np.var(b, ddof=1)) / 2)

def effect_label(d):
    d = abs(d)
    return 'пренебрежимо мал' if d < 0.2 else 'малый' if d < 0.5 else 'средний' if d < 0.8 else 'большой'

def decision(p, alpha=ALPHA):
    return 'отвергаем H0' if p < alpha else 'нет оснований отвергнуть H0'

results = []   # сюда собираем p-значения всех тестов для поправки на множественность
```

### 5.1. Одновыборочный t-тест

**Вопрос:** равен ли средний балл по математике 70 (условный «хороший» уровень, например целевой показатель школы)?

* $H_0: \mu = 70$
* $H_1: \mu \ne 70$ (двусторонняя)

Условия: наблюдения независимы (разные ученики); строгой нормальности нет, но при $n = 1000$ по ЦПТ
распределение выборочного среднего практически нормально, поэтому t-тест применим.

```python
mu0 = 70
t1, p1 = stats.ttest_1samp(x, mu0)
d1 = (x.mean() - mu0) / x.std(ddof=1)

# Бутстреп-проверка: сдвигаем выборку так, чтобы H0 была верна, и смотрим,
# как часто |t*| >= |t_набл|  (непараметрический аналог t-теста)
x0 = x - x.mean() + mu0
idx = rng.integers(0, n, size=(B, n))
bs = x0[idx]
t_boot = (bs.mean(axis=1) - mu0) / (bs.std(axis=1, ddof=1) / np.sqrt(n))
p1_boot = (np.sum(np.abs(t_boot) >= abs(t1)) + 1) / (B + 1)

print(f'H0: μ = {mu0}')
print(f'x̄ = {x.mean():.3f}, t = {t1:.4f}, df = {n-1}, p = {p1:.3e}')
print(f'95% ДИ для μ: ({ci95[0]:.3f}, {ci95[1]:.3f}) — {"не содержит" if not ci95[0] <= mu0 <= ci95[1] else "содержит"} {mu0}')
print(f'Бутстреп p-значение: {p1_boot:.5f} (минимально возможное при B={B}: {1/(B+1):.5f})')
print(f"Cohen's d = {d1:.3f} ({effect_label(d1)} эффект)")
print(f'Вывод: {decision(p1)}')
results.append(('Одновыборочный t (μ=70)', p1))
```

**Интерпретация.** p ≈ 10⁻¹⁵ ≪ 0.05, поэтому отвергаем H₀: средний балл по математике статистически значимо **ниже 70**. Это согласуется с доверительным интервалом: 95% ДИ (65.15; 67.03) не содержит 70. Бутстреп-проверка даёт тот же результат (ни одна из 10 000 бутстреп-статистик не оказалась экстремальнее наблюдаемой).

Статистически эффект огромен (t = −8.2), но практически он умеренный: средний ученик недобирает до цели около 3–5 баллов, Cohen's d = −0.26 (малый эффект).

### 5.2. Двухвыборочный t-тест

Сравниваем средний балл по математике в двух парах независимых групп:

1. **Пол:** $H_0: \mu_{male} = \mu_{female}$
2. **Подготовительный курс:** $H_0: \mu_{completed} = \mu_{none}$

Сначала тестом Левена проверяем равенство дисперсий и по его результату выбираем Стьюдента или Уэлча.
Для разности средних дополнительно строим бутстреп-ДИ (выборки ресэмплируются независимо внутри каждой группы).

```python
def two_sample_analysis(a, b, name_a, name_b, title):
    a, b = np.asarray(a, float), np.asarray(b, float)
    _, p_lev = stats.levene(a, b)
    equal_var = p_lev > ALPHA
    res = stats.ttest_ind(a, b, equal_var=equal_var)
    ci = res.confidence_interval(0.95)          # ДИ для разности средних (scipy >= 1.11)
    d = cohen_d_ind(a, b)

    # Бутстреп-ДИ для разности средних
    ia = rng.integers(0, len(a), size=(B, len(a)))
    ib = rng.integers(0, len(b), size=(B, len(b)))
    diff_boot = a[ia].mean(axis=1) - b[ib].mean(axis=1)
    boot_ci = np.percentile(diff_boot, [2.5, 97.5])

    print(f'=== {title} ===')
    print(f'{name_a}: n = {len(a)}, x̄ = {a.mean():.2f}, s = {a.std(ddof=1):.2f}')
    print(f'{name_b}: n = {len(b)}, x̄ = {b.mean():.2f}, s = {b.std(ddof=1):.2f}')
    print(f'Тест Левена: p = {p_lev:.4f} -> {"дисперсии равны, t-тест Стьюдента" if equal_var else "дисперсии различны, t-тест Уэлча"}')
    print(f'Разность средних = {a.mean() - b.mean():.3f}')
    print(f't = {res.statistic:.4f}, df = {res.df:.1f}, p = {res.pvalue:.3e}')
    print(f'95% t-ДИ разности:        ({ci.low:.3f}, {ci.high:.3f})')
    print(f'95% бутстреп-ДИ разности: ({boot_ci[0]:.3f}, {boot_ci[1]:.3f})')
    print(f"Cohen's d = {d:.3f} ({effect_label(d)} эффект)")
    print(f'Вывод: {decision(res.pvalue)}\n')
    return res.pvalue, diff_boot

male, female = df.loc[df.gender == 'male', 'math'], df.loc[df.gender == 'female', 'math']
done, none_ = df.loc[df.prep == 'completed', 'math'], df.loc[df.prep == 'none', 'math']

p2a, diff_gender = two_sample_analysis(male, female, 'male', 'female', 'math score: юноши vs девушки')
p2b, diff_prep = two_sample_analysis(done, none_, 'completed', 'none', 'math score: курс пройден vs не пройден')
results += [('Двухвыборочный t (пол)', p2a), ('Двухвыборочный t (курс)', p2b)]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 4.5))
for ax, diff, title in [(axes[0], diff_gender, 'male − female'), (axes[1], diff_prep, 'completed − none')]:
    ax.hist(diff, bins=50, color='lightseagreen', alpha=0.7, edgecolor='white')
    lo, hi = np.percentile(diff, [2.5, 97.5])
    ax.axvline(lo, color='red', ls='--'); ax.axvline(hi, color='red', ls='--', label='95% ДИ')
    ax.axvline(0, color='black', lw=2, label='0 (нет различий)')
    ax.set_title(f'Бутстреп-распределение разности средних: {title}')
    ax.set_xlabel('Разность средних math score'); ax.legend()
plt.tight_layout(); plt.show()
```

**Интерпретация.**

* **Пол.** Юноши в среднем набирают на **5.1 балла больше** девушек (95% ДИ: 3.2–7.0), p ≈ 9·10⁻⁸. Дисперсии групп равны (Левен p = 0.56), поэтому применён t-тест Стьюдента. Cohen's d = 0.34 — малый, но заметный эффект.
* **Подготовительный курс.** Прошедшие курс в среднем набирают на **5.6 балла больше** (95% ДИ: 3.7–7.6), p ≈ 1.5·10⁻⁸, d = 0.38 (малый-средний эффект).

В обоих случаях ДИ разности не содержит 0, что эквивалентно p < 0.05. Бутстреп-интервалы почти совпадают с t-интервалами. **Важно:** данные наблюдательные, а не экспериментальные — ученики сами решали, проходить ли курс, поэтому разница не обязательно означает, что курс *причина* роста баллов (возможно, курс выбирают более мотивированные ученики).

### 5.3. Парный t-тест

У каждого ученика есть два связанных измерения — баллы за **чтение** и за **письмо**.
Выборки зависимы (сильный ученик обычно хорошо пишет и читает), поэтому двухвыборочный тест здесь
некорректен — анализируем разности $d_i = \text{reading}_i - \text{writing}_i$.

* $H_0: \mu_d = 0$ (в среднем баллы за чтение и письмо одинаковы)
* $H_1: \mu_d \ne 0$

```python
reading = df['reading'].to_numpy(float)
writing = df['writing'].to_numpy(float)
diffs = reading - writing

t3, p3 = stats.ttest_rel(reading, writing)
t3_alt, p3_alt = stats.ttest_1samp(diffs, 0)        # то же самое через разности
d3 = diffs.mean() / diffs.std(ddof=1)               # Cohen's d для парных данных
ci_d = t_ci(diffs)
boot_d = bootstrap_ci(diffs, np.mean)
w_stat, p3_w = stats.wilcoxon(reading, writing)     # непараметрическая проверка

print(f'Корреляция reading–writing: r = {np.corrcoef(reading, writing)[0, 1]:.3f}')
print(f'Среднее разностей = {diffs.mean():.3f}, s_d = {diffs.std(ddof=1):.3f}')
print(f'Парный t-тест:              t = {t3:.4f}, df = {n-1}, p = {p3:.3e}')
print(f'Через разности (проверка):  t = {t3_alt:.4f}, p = {p3_alt:.3e}')
print(f'95% t-ДИ для μ_d:           ({ci_d[0]:.3f}, {ci_d[1]:.3f})')
print(f'95% бутстреп-ДИ для μ_d:    ({boot_d["percentile"][0]:.3f}, {boot_d["percentile"][1]:.3f})')
print(f'Критерий Уилкоксона:        W = {w_stat:.1f}, p = {p3_w:.3e}')
print(f"Cohen's d = {d3:.3f} ({effect_label(d3)} эффект)")
print(f'Вывод: {decision(p3)}')

# Для сравнения: что дал бы (неправильный здесь) тест для независимых выборок
p3_wrong = stats.ttest_ind(reading, writing).pvalue
print(f'\nДля сравнения — некорректный двухвыборочный t-тест: p = {p3_wrong:.4f}')
results.append(('Парный t (reading vs writing)', p3))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
axes[0].hist(diffs, bins=range(int(diffs.min()) - 1, int(diffs.max()) + 2), color='seagreen', alpha=0.7, edgecolor='white')
axes[0].axvline(0, color='red', ls='--', label='нулевая разность')
axes[0].axvline(diffs.mean(), color='black', label=f'среднее = {diffs.mean():.2f}')
axes[0].set_title('Разности reading − writing'); axes[0].legend()
stats.probplot(diffs, dist='norm', plot=axes[1])
axes[1].set_title('Q-Q график разностей')
plt.tight_layout(); plt.show()
```

**Интерпретация.** Баллы за чтение в среднем на **1.1 балла выше**, чем за письмо (95% ДИ: 0.83–1.40), p ≈ 2·10⁻¹⁴ — различие статистически значимо. Непараметрический критерий Уилкоксона подтверждает вывод, а Q-Q график показывает, что разности распределены почти нормально, так что условие парного t-теста выполнено.

Два важных наблюдения:

1. **Почему нужен именно парный тест.** Корреляция чтения и письма r = 0.955. Если (неправильно) применить тест для независимых выборок, получим p = 0.094 — различие *не обнаружится*, т.к. огромная межиндивидуальная изменчивость (s ≈ 15) «заглушит» маленькое систематическое различие. Парный тест убирает эту изменчивость (s_d = 4.5) и потому гораздо мощнее.
2. **Статистическая ≠ практическая значимость.** Разница в 1 балл из 100 практически ничтожна, хотя p крошечное. Это классический пример из п. 3.6 методички.

### 5.4. Критерий хи-квадрат

**(а) Согласие с нормальным распределением.** Проверяем, согласуется ли распределение `math score`
с нормальным законом с параметрами, оценёнными по выборке.

* Интервалы строим по квантилям нормального распределения → ожидаемые частоты примерно равны (~50 в каждом из 20 интервалов; после округления границ до полуцелых — от 26 до 71), условие $E_i \ge 5$ выполнено.
* Так как оценены 2 параметра ($\mu$, $\sigma$), число степеней свободы $k - 1 - 2$ (`ddof=2`).

> В шаблоне методички наблюдаемые частоты сравнивались с **равномерным** распределением по интервалам гистограммы — это проверка
> не нормальности, а равномерности, и для нормальных данных она заведомо отвергается. Здесь это исправлено.

```python
k = 20
mu_hat, sigma_hat = x.mean(), x.std(ddof=1)
inner_edges = stats.norm.ppf(np.linspace(0, 1, k + 1)[1:-1], mu_hat, sigma_hat)
# баллы целые, поэтому сдвигаем границы на 0.5, чтобы не было «привязки» значений к границе
inner_edges = np.floor(inner_edges) + 0.5
edges = np.concatenate([[-np.inf], inner_edges, [np.inf]])

observed = np.histogram(x, bins=edges)[0]
expected = n * np.diff(stats.norm.cdf(edges, mu_hat, sigma_hat))
expected *= observed.sum() / expected.sum()

chi2_gof, p4a = stats.chisquare(observed, expected, ddof=2)
print(f'H0: math score ~ N({mu_hat:.2f}, {sigma_hat:.2f}²)')
print(f'χ² = {chi2_gof:.3f}, df = {k - 1 - 2}, p = {p4a:.4f}')
print(f'Минимальная ожидаемая частота: {expected.min():.1f}')
print(f'Вывод: {decision(p4a)}')
results.append(('χ² согласие (нормальность)', p4a))

fig, ax = plt.subplots(figsize=(12, 4))
pos = np.arange(k)
ax.bar(pos - 0.2, observed, width=0.4, label='Наблюдаемые', color='steelblue')
ax.bar(pos + 0.2, expected, width=0.4, label='Ожидаемые (нормальное)', color='salmon')
ax.set_xticks(pos)
ax.set_xticklabels([f'≤{e:.0f}' if np.isfinite(e) else '>' for e in edges[1:]], rotation=45)
ax.set_title('Хи-квадрат: наблюдаемые и ожидаемые частоты по интервалам'); ax.legend()
plt.tight_layout(); plt.show()
```

**Интерпретация.** p = 0.64 — нет оснований отвергнуть гипотезу о нормальном распределении math score. Наблюдаемые частоты по интервалам близки к ожидаемым.

Обратите внимание на противоречие с тестом Шапиро–Уилка (p = 0.0001). Это не ошибка: критерий хи-квадрат «огрубляет» данные, группируя их в интервалы, и теряет чувствительность к тонким особенностям формы (например, к тяжести левого хвоста — все 8 выбросов попадают в один крайний интервал вместе с ~50 обычными значениями). Шапиро–Уилк использует все значения и намного мощнее. Итог: распределение **приближённо** нормально — отклонения есть, но они небольшие.

**(б) Независимость категориальных признаков.** Связан ли тип обеда (`lunch` — косвенный показатель
дохода семьи) с прохождением подготовительного курса (`prep`)? И связан ли тип обеда с тем,
**сдал ли ученик математику** (балл ≥ 50)?

* $H_0$: признаки независимы
* $H_1$: между признаками есть связь

Размер эффекта — коэффициент V Крамера.

```python
df['math_pass'] = np.where(df['math'] >= 50, 'pass', 'fail')

def chi2_independence(col1, col2):
    table = pd.crosstab(df[col1], df[col2])
    chi2, p, dof, exp = stats.chi2_contingency(table)   # для 2x2 — с поправкой Йетса
    v = stats.contingency.association(table.values, method='cramer')
    print(f'=== {col1} × {col2} ===')
    print(table, '\n')
    print('Ожидаемые частоты:\n', pd.DataFrame(exp, index=table.index, columns=table.columns).round(1), '\n')
    print(f'χ² = {chi2:.3f}, df = {dof}, p = {p:.4g}, V Крамера = {v:.3f}')
    print(f'Вывод: {decision(p)}\n')
    return p, table

p4b, tab_prep = chi2_independence('lunch', 'prep')
p4c, tab_pass = chi2_independence('lunch', 'math_pass')
results += [('χ² независимость (lunch × prep)', p4b), ('χ² независимость (lunch × pass)', p4c)]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
for ax, tab, title in [(axes[0], tab_prep, 'Доля прошедших курс'), (axes[1], tab_pass, 'Доля сдавших математику')]:
    (tab.div(tab.sum(axis=1), axis=0) * 100).plot(kind='bar', stacked=True, ax=ax, colormap='Set2', rot=0)
    ax.set_ylabel('%'); ax.set_title(f'{title} в зависимости от типа обеда')
    ax.legend(loc='lower right')
plt.tight_layout(); plt.show()
```

**Интерпретация.**

* **lunch × prep:** p = 0.64, V Крамера = 0.017 — связи нет. Доля прошедших подготовительный курс одинакова (~36–37%) у учеников с обычным и льготным обедом. Т.е. доступ к курсу не зависит от достатка семьи.
* **lunch × сдал математику:** p ≈ 4·10⁻²⁰, V = 0.29 — значимая связь средней силы. Среди учеников со льготным обедом не сдали математику **27%**, а с обычным — только **6%**. Социально-экономический фактор сильно связан с успеваемостью (опять же, это связь, а не доказанная причинность).

### 5.5. Поправка на множественные сравнения

Мы провели несколько тестов на одних данных, поэтому вероятность хотя бы одной ошибки первого рода
выше 5%. Применяем поправку Бонферрони и более мощный метод Холма.

```python
names = [r[0] for r in results]
pvals = np.array([r[1] for r in results])
m = len(pvals)

p_bonf = np.minimum(pvals * m, 1)

order = np.argsort(pvals)                                 # метод Холма
p_holm = np.empty(m)
running_max = 0
for rank, i in enumerate(order):
    running_max = max(running_max, (m - rank) * pvals[i])
    p_holm[i] = min(running_max, 1)

mc = pd.DataFrame({'Тест': names, 'p': pvals, 'p (Бонферрони)': p_bonf, 'p (Холм)': p_holm})
mc['Значим (Холм, α=0.05)'] = mc['p (Холм)'] < ALPHA
print(f'Число тестов m = {m}; без поправки P(хотя бы 1 ложное отвержение) ≤ 1 − 0.95^{m} = {1 - 0.95**m:.3f}')
print(mc.to_string(index=False, float_format=lambda v: f'{v:.3g}'))
```

**Интерпретация.** Проведено 7 тестов, без поправки вероятность хотя бы одного ложноположительного вывода могла достигать ~30%. Однако все значимые результаты имеют p порядка 10⁻⁸–10⁻²⁰, поэтому остаются значимыми и после самой консервативной поправки Бонферрони (α_скорр = 0.05/7 ≈ 0.007). Незначимые результаты остаются незначимыми. Выводы устойчивы к проблеме множественных сравнений.

При меньшем объёме выборки или сильной асимметрии различия между методами были бы заметнее.

**3. Какой тест наиболее адекватен данным.** Наиболее корректно применим **парный t-тест** (reading vs writing): его условие — нормальность разностей — выполнено лучше, чем нормальность самих баллов (Q-Q график разностей почти идеален), наблюдения действительно парные, а результат подтверждён непараметрическим критерием Уилкоксона. Также хорошо обоснованы **двухвыборочные t-тесты**: группы большие (n > 350), дисперсии равны по тесту Левена, бутстреп подтверждает результат. Одновыборочный тест корректен, но выбор μ₀ = 70 условен. Критерий хи-квадрат согласия наименее информативен: из-за группировки он менее чувствителен к отклонениям от нормальности, чем тест Шапиро–Уилка.

**Ограничения:** данные наблюдательные (нельзя делать причинные выводы о влиянии курса или дохода); набор, по-видимому, синтетический/учебный; баллы ограничены диапазоном 0–100 и принимают только целые значения.
