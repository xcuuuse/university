# Лабораторная работа 3. Доверительные интервалы и проверка гипотез

**Набор данных:** [Weight loss using diet](https://www.kaggle.com/datasets/tombenny/foodhabbits) (Kaggle, файл `foodDiet.csv`).
Исходный источник — учебные материалы University of Sheffield (statstutor): эксперимент, в котором 78 взрослых
были распределены на одну из трёх диет, и их вес измерялся до начала и через 6 недель.

| Признак | Тип | Описание |
|---|---|---|
| `Person` | идентификатор | номер участника |
| `gender` | категориальный | пол (0 — женщина, 1 — мужчина), **есть пропуски** |
| `Age` | числовой | возраст, лет |
| `Height` | числовой | рост, см |
| `pre.weight` | непрерывный | вес до диеты, кг |
| `Diet` | категориальный | тип диеты (1, 2, 3) |
| `weight6weeks` | непрерывный | вес через 6 недель, кг |

**Почему этот набор подходит под задание:**

* есть естественные **парные измерения** «до / после» — идеальный случай для парного t-теста;
* по ним строится непрерывный признак **потеря веса** (`weight_loss = pre.weight − weight6weeks`) — основной признак для ДИ;
* есть **группы** (тип диеты, пол) для двухвыборочного t-теста;
* есть **категориальные признаки** для критерия хи-квадрат;
* выборка **небольшая** (n ≈ 76–78, в группах по ~25 человек) — на ней хорошо видно, когда бутстреп и параметрические методы расходятся;
* в данных есть **настоящие пропуски**, которые нужно осмысленно обработать.

**Основной исследуемый признак:** `weight_loss` — потеря веса за 6 недель, кг.

## 1. Подключение библиотек

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid')
plt.rcParams['figure.dpi'] = 100

ALPHA = 0.05
B = 10_000                       # число бутстреп-репликаций (требование: B >= 5000)
rng = np.random.default_rng(42)  # генератор случайных чисел для воспроизводимости

def qq_plot(data, ax, title):
    # Q-Q график с русскими подписями осей
    stats.probplot(data, dist='norm', plot=ax)
    ax.set_title(title)
    ax.set_xlabel('Теоретические квантили N(0, 1)')
    ax.set_ylabel('Упорядоченные значения выборки')
```

### 2.1. Пропуски и дубликаты

```python
print('Пропуски по столбцам:')
print(raw.isna().sum())
print(f'\nПолных дубликатов строк: {raw.duplicated().sum()}')
print(f'Дубликатов номеров участников: {raw["Person"].duplicated().sum()}')
print('\nСтроки с пропусками:')
raw[raw.isna().any(axis=1)]
```

**Анализ пропусков.** Пропущен только пол у двух участников (№ 25 и 26). При этом у обоих вес через 6 недель
**в точности равен** исходному (60 → 60.0 и 103 → 103.0), тогда как у всех остальных вес записан с точностью
до 0.1 кг и изменился. Скорее всего, это участники, выбывшие из эксперимента, для которых итоговое
измерение не проводилось и было просто скопировано из начального.

Варианты обработки:

1. *Заполнить пол модой* — неверно: восстанавливать пол «наугад» бессмысленно, а главная проблема этих строк не в поле, а в недостоверном итоговом весе.
2. *Удалить только из анализов по полу* — оставит в данных два фиктивных «нулевых» результата и занизит средний эффект диеты.
3. **Удалить обе строки целиком** — выбранный вариант. Это потеря 2 из 78 наблюдений (2.6%), не критично.

Ниже для проверки устойчивости приведено, как изменилась бы средняя потеря веса, если бы эти строки оставили.

```python
# Переименуем столбцы в удобный вид
df = raw.rename(columns={'Person': 'person', 'gender': 'gender', 'Age': 'age', 'Height': 'height',
                         'pre.weight': 'pre_weight', 'Diet': 'diet', 'weight6weeks': 'weight_6w'})

loss_with_na = df['pre_weight'] - df['weight_6w']
df = df.dropna().reset_index(drop=True)

# Новые признаки
df['weight_loss'] = df['pre_weight'] - df['weight_6w']                 # потеря веса, кг
df['loss_pct'] = df['weight_loss'] / df['pre_weight'] * 100            # потеря веса, % от исходного
df['bmi_pre'] = df['pre_weight'] / (df['height'] / 100) ** 2           # ИМТ до диеты
df['gender'] = df['gender'].astype(int).map({0: 'Женщины', 1: 'Мужчины'}).astype('category')
df['diet'] = df['diet'].astype(int).map({1: 'Диета 1', 2: 'Диета 2', 3: 'Диета 3'}).astype('category')

print(f'После удаления: {df.shape[0]} наблюдений, пропусков: {df.isna().sum().sum()}')
print(f'Средняя потеря веса с «подозрительными» строками: {loss_with_na.mean():.3f} кг (n = {len(loss_with_na)})')
print(f'Средняя потеря веса без них:                      {df["weight_loss"].mean():.3f} кг (n = {len(df)})')
print('\nРаспределение по группам:')
print(pd.crosstab(df['gender'], df['diet'], margins=True, margins_name='Всего'))
```

### 2.2. Описательные статистики

```python
num_cols = ['age', 'height', 'pre_weight', 'weight_6w', 'weight_loss', 'loss_pct']
desc = df[num_cols].describe().T
desc['median'] = df[num_cols].median()
desc['IQR'] = desc['75%'] - desc['25%']
desc['skew'] = df[num_cols].skew()
desc['kurtosis'] = df[num_cols].kurt()
desc.round(2)
```

```python
x = df['weight_loss'].to_numpy(float)   # основной признак
n = len(x)

print('Потеря веса по типу диеты:')
print(df.groupby('diet', observed=True)['weight_loss'].agg(['count', 'mean', 'median', 'std']).round(3))
print('\nПотеря веса по полу:')
print(df.groupby('gender', observed=True)['weight_loss'].agg(['count', 'mean', 'median', 'std']).round(3))

# Выбросы по правилу 1.5*IQR
for col in ['weight_loss', 'height', 'pre_weight']:
    q1, q3 = df[col].quantile([0.25, 0.75])
    lo, hi = q1 - 1.5 * (q3 - q1), q3 + 1.5 * (q3 - q1)
    out = sorted(df.loc[(df[col] < lo) | (df[col] > hi), col].round(1).tolist())
    print(f'\nВыбросы {col} (границы [{lo:.1f}; {hi:.1f}]): {out}')

print(f'\nУчастников, набравших вес: {(x < 0).sum()}  ->  {sorted(np.round(x[x < 0], 1).tolist())}')
```

**Комментарий к описательным статистикам.**

* Средняя потеря веса за 6 недель — **3.95 кг** (медиана 3.70 кг), что составляет около **5.5%** исходного веса. Среднее и медиана близки, асимметрия ≈ 0 — распределение симметричное.
* Разброс большой: от набора 2.1 кг до потери 9.2 кг (s = 2.51 кг). Четверо участников за 6 недель **набрали** вес — это нормальная индивидуальная изменчивость, а не ошибки данных.
* По группам: на диете 3 средняя потеря (5.15 кг) заметно больше, чем на диетах 1 и 2 (≈ 3.3 кг). Между мужчинами и женщинами различие мало (4.02 против 3.89 кг).
* Выбросов в потере веса нет. В росте по правилу 1.5·IQR формально выделяется 6 значений (141 см и 194–201 см) — это редкие, но физически возможные значения. Рост в проверке гипотез не используется, поэтому оставляем их без изменений.

## 3. Параметрические доверительные интервалы

**Для среднего** (дисперсия неизвестна) — t-интервал:
$\bar{x} \pm t_{1-\alpha/2,\,n-1}\cdot\dfrac{s}{\sqrt n}$.

**Для дисперсии** — интервал на основе $\chi^2$-распределения (требует нормальности данных):
$\left(\dfrac{(n-1)s^2}{\chi^2_{1-\alpha/2,\,n-1}},\ \dfrac{(n-1)s^2}{\chi^2_{\alpha/2,\,n-1}}\right)$.

```python
def t_ci(data, alpha=0.05):
    # Параметрический ДИ для среднего на основе t-распределения
    data = np.asarray(data, dtype=float)
    k = len(data)
    m, se = data.mean(), stats.sem(data)
    t_crit = stats.t.ppf(1 - alpha / 2, df=k - 1)
    return m - t_crit * se, m + t_crit * se

def var_ci(data, alpha=0.05):
    # Параметрический ДИ для дисперсии на основе chi2-распределения
    data = np.asarray(data, dtype=float)
    k, s2 = len(data), np.var(data, ddof=1)
    return ((k - 1) * s2 / stats.chi2.ppf(1 - alpha / 2, k - 1),
            (k - 1) * s2 / stats.chi2.ppf(alpha / 2, k - 1))

ci95, ci99, vci95 = t_ci(x, 0.05), t_ci(x, 0.01), var_ci(x, 0.05)

print(f'n = {n}, выборочное среднее = {x.mean():.3f} кг, SE = {stats.sem(x):.3f}')
print(f'95% t-ДИ для среднего: ({ci95[0]:.3f}; {ci95[1]:.3f}), ширина = {ci95[1] - ci95[0]:.3f}')
print(f'99% t-ДИ для среднего: ({ci99[0]:.3f}; {ci99[1]:.3f}), ширина = {ci99[1] - ci99[0]:.3f}')
print(f'\nВыборочная дисперсия = {x.var(ddof=1):.3f} (s = {x.std(ddof=1):.3f})')
print(f'95% χ²-ДИ для дисперсии:             ({vci95[0]:.3f}; {vci95[1]:.3f})')
print(f'95% ДИ для стандартного отклонения:  ({np.sqrt(vci95[0]):.3f}; {np.sqrt(vci95[1]):.3f})')
print('\nПроверка через scipy:', tuple(np.round(stats.t.interval(0.95, n - 1, loc=x.mean(), scale=stats.sem(x)), 3)))
```

**Интерпретация.** С доверительной вероятностью 95% средняя потеря веса за 6 недель в генеральной совокупности лежит в интервале **(3.37; 4.52) кг**, с вероятностью 99% — в **(3.19; 4.71) кг**. 99%-интервал шире на треть (1.52 против 1.15 кг): за большую надёжность платим точностью. Оба интервала целиком лежат выше нуля — диеты в среднем действительно приводят к похудению.

Стандартное отклонение потери веса с вероятностью 95% лежит в пределах (2.16; 2.98) кг — индивидуальные результаты сильно варьируют вокруг среднего.

## 4. Бутстреп-доверительные интервалы

Улучшения по сравнению с кодом из методички:

* бутстреп **векторизован** (все $B$ псевдовыборок генерируются одной матрицей индексов) — в десятки раз быстрее цикла;
* нормальный интервал центрируется на оценке по **исходной** выборке: $\hat\theta \pm z_{1-\alpha/2}\cdot SE_{boot}$, как в формуле из теории;
* выводится оценка **смещения** бутстрепа и дополнительно строится **BCa-интервал** (через `scipy.stats.bootstrap`);
* для бутстреп-ДИ дисперсии — сравнение с χ²-интервалом.

```python
def bootstrap_ci(data, statistic, n_bootstrap=B, ci=95, rng=rng):
    # Возвращает нормальный и процентильный бутстреп-ДИ и сами бутстреп-оценки.
    # statistic должна принимать параметр axis (np.mean, np.median, np.var и т.п.).
    data = np.asarray(data, dtype=float)
    k = len(data)
    idx = rng.integers(0, k, size=(n_bootstrap, k))   # B псевдовыборок с возвращением
    estimates = statistic(data[idx], axis=1)

    theta_hat = statistic(data)
    se_boot = estimates.std(ddof=1)
    a = 1 - ci / 100
    z = stats.norm.ppf(1 - a / 2)
    return {'theta': theta_hat, 'se': se_boot, 'bias': estimates.mean() - theta_hat,
            'normal': (theta_hat - z * se_boot, theta_hat + z * se_boot),
            'percentile': tuple(np.percentile(estimates, [100 * a / 2, 100 * (1 - a / 2)])),
            'estimates': estimates}

def bca_ci(data, statistic, n_bootstrap=B, ci=95):
    # BCa-интервал: scipy сам оценивает поправку смещения z0 и ускорение a (jackknife)
    res = stats.bootstrap((np.asarray(data, dtype=float),), statistic, n_resamples=n_bootstrap,
                          confidence_level=ci / 100, method='BCa', random_state=42)
    return res.confidence_interval.low, res.confidence_interval.high

var_ddof1 = lambda a, axis=None: np.var(a, ddof=1, axis=axis)

boot_mean = bootstrap_ci(x, np.mean)
boot_median = bootstrap_ci(x, np.median)
boot_var = bootstrap_ci(x, var_ddof1)
bca_mean, bca_median, bca_var = bca_ci(x, np.mean), bca_ci(x, np.median), bca_ci(x, var_ddof1)

for name, r, bca in [('Среднее', boot_mean, bca_mean), ('Медиана', boot_median, bca_median),
                     ('Дисперсия', boot_var, bca_var)]:
    print(f'=== {name}: оценка = {r["theta"]:.3f}, SE_boot = {r["se"]:.3f}, смещение = {r["bias"]:+.4f}')
    print(f'  Нормальный 95%:    ({r["normal"][0]:.3f}; {r["normal"][1]:.3f})')
    print(f'  Процентильный 95%: ({r["percentile"][0]:.3f}; {r["percentile"][1]:.3f})')
    print(f'  BCa 95%:           ({bca[0]:.3f}; {bca[1]:.3f})')
print(f'\nДля сравнения: χ²-ДИ для дисперсии ({vci95[0]:.3f}; {vci95[1]:.3f})')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

ax = axes[0]
est = boot_mean['estimates']
ax.hist(est, bins=50, density=True, alpha=0.6, color='skyblue', edgecolor='white')
g = np.linspace(est.min(), est.max(), 300)
ax.plot(g, stats.norm.pdf(g, boot_mean['theta'], boot_mean['se']), 'k-', lw=1.5, label='Нормальная аппроксимация')
ax.axvline(boot_mean['theta'], color='black', ls=':', label=f'Исходное среднее ({boot_mean["theta"]:.2f})')
for v in boot_mean['percentile']:
    ax.axvline(v, color='red', ls='--')
for v in ci95:
    ax.axvline(v, color='green', ls='-.')
ax.plot([], [], 'r--', label='Процентильный 95% ДИ')
ax.plot([], [], 'g-.', label='Параметрический 95% t-ДИ')
ax.set_title(f'Бутстреп-распределение среднего (B = {B})')
ax.set_xlabel('Средняя потеря веса, кг'); ax.set_ylabel('Плотность'); ax.legend(fontsize=9)

ax = axes[1]
est = boot_median['estimates']
ax.hist(est, bins=40, density=True, alpha=0.6, color='plum', edgecolor='white')
ax.axvline(boot_median['theta'], color='black', ls=':', label=f'Исходная медиана ({boot_median["theta"]:.2f})')
for v in boot_median['percentile']:
    ax.axvline(v, color='red', ls='--')
ax.plot([], [], 'r--', label='Процентильный 95% ДИ')
ax.set_title(f'Бутстреп-распределение медианы (B = {B})')
ax.set_xlabel('Медиана потери веса, кг'); ax.set_ylabel('Плотность'); ax.legend(fontsize=9)

plt.tight_layout()
plt.show()
```

### 4.1. Сравнение параметрических и бутстреп-интервалов

```python
rows = [
    ('Среднее', 'Параметрический (t)', *ci95),
    ('Среднее', 'Бутстреп нормальный', *boot_mean['normal']),
    ('Среднее', 'Бутстреп процентильный', *boot_mean['percentile']),
    ('Среднее', 'Бутстреп BCa', *bca_mean),
    ('Медиана', 'Бутстреп процентильный', *boot_median['percentile']),
    ('Медиана', 'Бутстреп BCa', *bca_median),
]
ci_table = pd.DataFrame(rows, columns=['Статистика', 'Метод', 'Нижняя', 'Верхняя'])
ci_table['Ширина'] = ci_table['Верхняя'] - ci_table['Нижняя']
print(f'Параметрическая SE = s/√n = {stats.sem(x):.4f};  бутстреп SE среднего = {boot_mean["se"]:.4f}')
print(f'Отношение SE_boot / SE_param = {boot_mean["se"] / stats.sem(x):.4f}  (теоретически ≈ √((n−1)/n) = {np.sqrt((n - 1) / n):.4f})')
ci_table.round(3)
```

### 4.2. Малые выборки: ДИ для каждой диеты отдельно

В каждой группе всего 24–27 человек — именно здесь, по методичке, бутстреп может давать смещённые
(слишком узкие) интервалы. Сравним методы внутри групп.

```python
rows = []
for g, sub in df.groupby('diet', observed=True)['weight_loss']:
    sub = sub.to_numpy(float)
    tci = t_ci(sub)
    r = bootstrap_ci(sub, np.mean)
    bca = bca_ci(sub, np.mean)
    rows.append([g, len(sub), sub.mean(), *tci, *r['percentile'], *bca,
                 (r['percentile'][1] - r['percentile'][0]) / (tci[1] - tci[0])])
group_ci = pd.DataFrame(rows, columns=['Диета', 'n', 'Среднее', 't: нижн.', 't: верх.',
                                       'Перц.: нижн.', 'Перц.: верх.', 'BCa: нижн.', 'BCa: верх.',
                                       'Ширина перц. / ширина t'])
group_ci.round(3)
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
offs = {'t-интервал': (-0.2, 't: нижн.', 't: верх.', 'tab:green'),
        'Процентильный': (0, 'Перц.: нижн.', 'Перц.: верх.', 'tab:red'),
        'BCa': (0.2, 'BCa: нижн.', 'BCa: верх.', 'tab:purple')}
for lab, (o, lo, hi, c) in offs.items():
    ax.errorbar(group_ci['Среднее'], np.arange(3) + o,
                xerr=[group_ci['Среднее'] - group_ci[lo], group_ci[hi] - group_ci['Среднее']],
                fmt='o', capsize=4, color=c, label=lab)
ax.set_yticks(range(3)); ax.set_yticklabels(group_ci['Диета']); ax.invert_yaxis()
ax.set_xlabel('Средняя потеря веса, кг'); ax.set_title('95% ДИ средней потери веса по диетам')
ax.legend(); plt.tight_layout(); plt.show()
```

**Сравнение и объяснение различий.**

* **Среднее по всей выборке (n = 76).** Все четыре интервала почти совпадают, но бутстреп-интервалы систематически **чуть уже** t-интервала (1.12 против 1.15 кг). Причины две: (1) бутстреп оценивает SE как s·√((n−1)/n), что немного меньше s/√n (отношение 0.992 — ровно теоретическое значение); (2) бутстреп использует квантили нормального распределения (z = 1.96), а t-интервал — более «тяжёлые» квантили t-распределения (t = 1.99 при df = 75). Бутстреп «не знает», что дисперсия тоже оценена по выборке.
* **Малые группы (n = 24–27).** Здесь эффект заметнее: процентильный бутстреп-ДИ на **~7% уже** t-интервала для каждой диеты. Это прямая иллюстрация замечания из методички: при малых выборках простой бутстреп даёт слишком оптимистичные (узкие) интервалы, и их реальная вероятность покрытия ниже заявленных 95%. При нормальных данных t-интервал здесь надёжнее.
* **Медиана.** Процентильный 95% ДИ — **(3.2; 4.3) кг**. Он близок по ширине к интервалу для среднего. Бутстреп-распределение медианы неровное (медиана выборки принимает лишь значения из исходных данных), поэтому для неё подходят процентильный метод или BCa, но не нормальный.
* **Дисперсия.** Здесь методы расходятся сильнее всего: χ²-ДИ (4.67; 8.90), процентильный бутстреп (4.51; 8.05), BCa (4.80; 8.44). Распределение выборочной дисперсии асимметрично (длинный правый хвост), и обычные бутстреп-интервалы его недооценивают; BCa, учитывающий асимметрию, ближе к точному χ²-интервалу. Поскольку данные нормальны, χ²-интервал здесь предпочтительнее.

## 5. Проверка гипотез

Уровень значимости $\alpha = 0.05$. Помимо p-значения для каждого теста приводится доверительный интервал
для эффекта и **размер эффекта** (Cohen's d, V Крамера); в конце — поправка на множественные сравнения.

```python
def cohen_d_ind(a, b):
    # Cohen's d для независимых выборок (формула из методички)
    return (np.mean(a) - np.mean(b)) / np.sqrt((np.var(a, ddof=1) + np.var(b, ddof=1)) / 2)

def effect_label(d):
    d = abs(d)
    return 'пренебрежимо малый' if d < 0.2 else 'малый' if d < 0.5 else 'средний' if d < 0.8 else 'большой'

def decision(p, alpha=ALPHA):
    return 'отвергаем H0' if p < alpha else 'нет оснований отвергнуть H0'

results = []   # p-значения всех тестов для поправки на множественность
```

### 5.1. Одновыборочный t-тест

**Вопрос:** превышает ли средняя потеря веса «безопасный» темп похудения?
Часто рекомендуемый темп — около **0.5 кг в неделю**, т.е. **3 кг за 6 недель**.

* $H_0: \mu = 3$ кг
* $H_1: \mu \ne 3$ кг (двусторонняя альтернатива)

Условия: наблюдения независимы (разные люди), потеря веса распределена приближённо нормально (п. 2.3).

```python
mu0 = 3.0
t1, p1 = stats.ttest_1samp(x, mu0)
d1 = (x.mean() - mu0) / x.std(ddof=1)

# Бутстреп-проверка: сдвигаем выборку так, чтобы H0 была верна, и считаем долю |t*| >= |t_набл|
x0 = x - x.mean() + mu0
bs = x0[rng.integers(0, n, size=(B, n))]
t_boot = (bs.mean(axis=1) - mu0) / (bs.std(axis=1, ddof=1) / np.sqrt(n))
p1_boot = (np.sum(np.abs(t_boot) >= abs(t1)) + 1) / (B + 1)

print(f'H0: μ = {mu0} кг')
print(f'x̄ = {x.mean():.3f}, t = {t1:.4f}, df = {n - 1}, p = {p1:.4f}')
print(f'95% ДИ для μ: ({ci95[0]:.3f}; {ci95[1]:.3f}) — {"не содержит" if not ci95[0] <= mu0 <= ci95[1] else "содержит"} {mu0}')
print(f'Бутстреп p-значение: {p1_boot:.4f}')
print(f"Cohen's d = {d1:.3f} ({effect_label(d1)} эффект)")
print(f'Вывод: {decision(p1)}')
results.append(('Одновыборочный t (μ = 3 кг)', p1))
```

**Интерпретация.** p = 0.0015 < 0.05, поэтому отвергаем H₀: средняя потеря веса статистически значимо **больше 3 кг** за 6 недель, т.е. в среднем превышает темп 0.5 кг/неделю. Это согласуется с 95% ДИ (3.37; 4.52), который не содержит 3. Бутстреп-проверка гипотезы дала практически то же p-значение (0.002).

Практически превышение составляет около 1 кг (≈ 0.16 кг/неделю сверх нормы), Cohen's d = 0.38 — малый эффект: участники худели чуть быстрее рекомендуемого темпа, но не экстремально.

### 5.2. Двухвыборочный t-тест

**(а) Главный вопрос исследования — различаются ли диеты.** По описательным статистикам лидирует диета 3.
Сравним её с диетой 1:
$H_0: \mu_{\text{диета 3}} = \mu_{\text{диета 1}}$, $H_1: \mu_{\text{диета 3}} \ne \mu_{\text{диета 1}}$.

**(б) Пол:** $H_0: \mu_{\text{муж}} = \mu_{\text{жен}}$.

Сначала тест Левена на равенство дисперсий → выбор между t-тестом Стьюдента и Уэлча.
Для разности средних дополнительно строится бутстреп-ДИ (ресэмплинг независимо внутри каждой группы)
и проверяется результат непараметрическим критерием Манна–Уитни.

```python
def two_sample_analysis(a, b, name_a, name_b, title):
    a, b = np.asarray(a, float), np.asarray(b, float)
    _, p_lev = stats.levene(a, b)
    equal_var = p_lev > ALPHA
    res = stats.ttest_ind(a, b, equal_var=equal_var)
    ci = res.confidence_interval(0.95)
    d = cohen_d_ind(a, b)
    p_mw = stats.mannwhitneyu(a, b).pvalue

    diff_boot = (a[rng.integers(0, len(a), size=(B, len(a)))].mean(axis=1)
                 - b[rng.integers(0, len(b), size=(B, len(b)))].mean(axis=1))
    boot_ci = np.percentile(diff_boot, [2.5, 97.5])

    print(f'=== {title} ===')
    print(f'{name_a}: n = {len(a)}, x̄ = {a.mean():.3f}, s = {a.std(ddof=1):.3f}')
    print(f'{name_b}: n = {len(b)}, x̄ = {b.mean():.3f}, s = {b.std(ddof=1):.3f}')
    print(f'Тест Левена: p = {p_lev:.4f} -> '
          f'{"дисперсии равны, t-тест Стьюдента" if equal_var else "дисперсии различны, t-тест Уэлча"}')
    print(f'Разность средних = {a.mean() - b.mean():.3f} кг')
    print(f't = {res.statistic:.4f}, df = {res.df:.1f}, p = {res.pvalue:.4f}')
    print(f'95% t-ДИ разности:        ({ci.low:.3f}; {ci.high:.3f})')
    print(f'95% бутстреп-ДИ разности: ({boot_ci[0]:.3f}; {boot_ci[1]:.3f})')
    print(f'Манна–Уитни: p = {p_mw:.4f}')
    print(f"Cohen's d = {d:.3f} ({effect_label(d)} эффект)")
    print(f'Вывод: {decision(res.pvalue)}\n')
    return res.pvalue, diff_boot

wl = lambda col, val: df.loc[df[col] == val, 'weight_loss']

p2a, diff_diet = two_sample_analysis(wl('diet', 'Диета 3'), wl('diet', 'Диета 1'),
                                     'Диета 3', 'Диета 1', 'Потеря веса: диета 3 vs диета 1')
p2b, diff_gender = two_sample_analysis(wl('gender', 'Мужчины'), wl('gender', 'Женщины'),
                                       'Мужчины', 'Женщины', 'Потеря веса: мужчины vs женщины')
results += [('Двухвыборочный t (диета 3 vs 1)', p2a), ('Двухвыборочный t (пол)', p2b)]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 4.5))
for ax, diff, title in [(axes[0], diff_diet, 'диета 3 − диета 1'), (axes[1], diff_gender, 'мужчины − женщины')]:
    ax.hist(diff, bins=50, color='lightseagreen', alpha=0.7, edgecolor='white')
    lo, hi = np.percentile(diff, [2.5, 97.5])
    ax.axvline(lo, color='red', ls='--'); ax.axvline(hi, color='red', ls='--', label='95% ДИ')
    ax.axvline(0, color='black', lw=2, label='0 (нет различий)')
    ax.set_title(f'Бутстреп-распределение разности средних:\n{title}')
    ax.set_xlabel('Разность средней потери веса, кг'); ax.set_ylabel('Частота'); ax.legend()
plt.tight_layout(); plt.show()
```

**Дополнительно: однофакторный дисперсионный анализ (ANOVA).** Диет три, и попарные t-тесты «выбирают»
сравнение уже после взгляда на данные. Корректнее сначала проверить общую гипотезу
$H_0: \mu_1 = \mu_2 = \mu_3$, а затем — все три попарных сравнения с поправкой Холма.

```python
groups = [g.to_numpy(float) for _, g in df.groupby('diet', observed=True)['weight_loss']]
f_stat, p_anova = stats.f_oneway(*groups)
_, p_lev3 = stats.levene(*groups)
print(f'Тест Левена (3 группы): p = {p_lev3:.4f}')
print(f'ANOVA: F = {f_stat:.3f}, p = {p_anova:.4f} -> {decision(p_anova)}')
print(f'Краскел–Уоллис (непараметрический аналог): p = {stats.kruskal(*groups).pvalue:.4f}\n')

pairs = [('Диета 1', 'Диета 2'), ('Диета 1', 'Диета 3'), ('Диета 2', 'Диета 3')]
pair_p = [stats.ttest_ind(wl('diet', a), wl('diet', b)).pvalue for a, b in pairs]
order = np.argsort(pair_p); holm = np.empty(3); run = 0
for rank, i in enumerate(order):
    run = max(run, (3 - rank) * pair_p[i]); holm[i] = min(run, 1)
print(pd.DataFrame({'Сравнение': [f'{a} vs {b}' for a, b in pairs],
                    'Разность средних': [wl('diet', a).mean() - wl('diet', b).mean() for a, b in pairs],
                    'p': pair_p, 'p (Холм)': holm}).round(4).to_string(index=False))
```

**Интерпретация.**

* **Диета 3 vs диета 1.** На диете 3 участники в среднем сбросили на **1.85 кг больше** (95% ДИ: 0.54–3.16 кг), p = 0.007. Дисперсии равны (Левен p = 0.36), применён t-тест Стьюдента. Бутстреп-ДИ (0.57–3.06) и критерий Манна–Уитни (p = 0.007) подтверждают вывод. Cohen's d = 0.80 — эффект на границе среднего и большого: по меркам диетологии разница почти в 2 кг за 6 недель существенна.
* **Пол.** Различия нет: разность 0.12 кг, p = 0.83, ДИ (−1.04; 1.29) содержит 0 и практически симметричен вокруг него, d = 0.05. Диеты работают одинаково для мужчин и женщин.
* **ANOVA** подтверждает, что диеты различаются в целом (F = 5.38, p = 0.007). Попарные сравнения с поправкой Холма: диета 3 лучше и диеты 1 (p = 0.020), и диеты 2 (p = 0.020), а диеты 1 и 2 между собой не различаются (p = 0.96).

### 5.3. Парный t-тест

Вес каждого участника измерен дважды — до диеты и через 6 недель. Это **зависимые** выборки, поэтому
анализируем разности $d_i = \text{вес до}_i - \text{вес после}_i$ (это и есть `weight_loss`).

* $H_0: \mu_d = 0$ — в среднем вес не изменился
* $H_1: \mu_d \ne 0$

Проверим по всей выборке и отдельно для каждой диеты.

```python
pre = df['pre_weight'].to_numpy(float)
post = df['weight_6w'].to_numpy(float)
diffs = pre - post

t3, p3 = stats.ttest_rel(pre, post)
t3_alt, p3_alt = stats.ttest_1samp(diffs, 0)
d3 = diffs.mean() / diffs.std(ddof=1)          # Cohen's d для парных данных
boot_d = bootstrap_ci(diffs, np.mean)
w_stat, p3_w = stats.wilcoxon(pre, post)

print(f'Вес до: {pre.mean():.2f} кг, после: {post.mean():.2f} кг; корреляция до/после r = {np.corrcoef(pre, post)[0, 1]:.3f}')
print(f'Среднее разностей = {diffs.mean():.3f} кг, s_d = {diffs.std(ddof=1):.3f}')
print(f'Парный t-тест:             t = {t3:.4f}, df = {n - 1}, p = {p3:.3e}')
print(f'Через разности (проверка): t = {t3_alt:.4f}, p = {p3_alt:.3e}')
print(f'95% t-ДИ для μ_d:          ({ci95[0]:.3f}; {ci95[1]:.3f})')
print(f'95% бутстреп-ДИ для μ_d:   ({boot_d["percentile"][0]:.3f}; {boot_d["percentile"][1]:.3f})')
print(f'Критерий Уилкоксона:       W = {w_stat:.1f}, p = {p3_w:.3e}')
print(f"Cohen's d = {d3:.3f} ({effect_label(d3)} эффект)")
print(f'Вывод: {decision(p3)}')

p3_wrong = stats.ttest_ind(pre, post).pvalue
print(f'\nДля сравнения — некорректный тест для независимых выборок: p = {p3_wrong:.4f}')

print('\nПарный t-тест по каждой диете:')
for g, sub in df.groupby('diet', observed=True):
    r = stats.ttest_rel(sub['pre_weight'], sub['weight_6w'])
    print(f'  {g}: n = {len(sub)}, средняя потеря = {sub["weight_loss"].mean():.2f} кг, t = {r.statistic:.2f}, p = {r.pvalue:.2e}')
results.append(('Парный t (до vs после)', p3))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 4.5))

ax = axes[0]
for i in range(n):
    ax.plot([0, 1], [pre[i], post[i]], color='gray', alpha=0.35)
ax.plot([0, 1], [pre.mean(), post.mean()], 'o-', color='red', lw=3, label='Средние')
ax.set_xticks([0, 1]); ax.set_xticklabels(['До диеты', 'Через 6 недель'])
ax.set_ylabel('Вес, кг'); ax.set_title('Изменение веса каждого участника'); ax.legend()

ax = axes[1]
ax.hist(diffs, bins=15, color='seagreen', alpha=0.7, edgecolor='white')
ax.axvline(0, color='red', ls='--', label='нулевая разность')
ax.axvline(diffs.mean(), color='black', label=f'среднее = {diffs.mean():.2f}')
ax.set_title('Разности (до − после)'); ax.set_xlabel('кг'); ax.set_ylabel('Частота'); ax.legend()

qq_plot(diffs, axes[2], 'Q-Q график разностей')
plt.tight_layout(); plt.show()
```

**Интерпретация.** В среднем вес снизился на **3.95 кг** (95% ДИ: 3.37–4.52 кг), t = 13.7, p ≈ 4·10⁻²² — изменение веса высоко значимо. Критерий Уилкоксона подтверждает вывод, а Q-Q график показывает, что разности распределены нормально, т.е. условие парного t-теста выполнено. Cohen's d = 1.58 — очень большой эффект. Значимое снижение веса наблюдается на **каждой** из трёх диет по отдельности.

**Почему тест именно парный.** Корреляция веса до и после r = 0.95: тяжёлые люди остаются тяжёлыми. Тест для независимых выборок сравнивал бы средние на фоне огромного межиндивидуального разброса (s ≈ 8 кг) и дал p = 0.003 — значимо, но в 10¹⁹ раз «слабее». Парный тест убирает индивидуальные различия (разброс разностей s_d = 2.5 кг) и поэтому гораздо мощнее.

Ограничение: в исследовании нет контрольной группы без диеты, поэтому строго говоря тест показывает, что вес **изменился**, но не доказывает, что причина — именно диета.

### 5.4. Критерий хи-квадрат

**(а) Согласие с нормальным распределением.** Проверяем, согласуется ли распределение потери веса
с нормальным законом с параметрами, оценёнными по выборке.

* Интервалы строим по квантилям нормального распределения → ожидаемые частоты равны ($E_i = n/k \approx 9.5 \ge 5$).
* Оценены 2 параметра ($\mu$, $\sigma$), поэтому степеней свободы $k - 1 - 2$ (`ddof=2`).

> В шаблоне из методички наблюдаемые частоты сравнивались с **равномерным** распределением по интервалам гистограммы —
> это проверка не нормальности, а равномерности. Здесь это исправлено.

```python
k = 8
mu_hat, sigma_hat = x.mean(), x.std(ddof=1)
edges = np.concatenate([[-np.inf], stats.norm.ppf(np.arange(1, k) / k, mu_hat, sigma_hat), [np.inf]])
observed = np.histogram(x, bins=edges)[0]
expected = np.full(k, n / k)

chi2_gof, p4a = stats.chisquare(observed, expected, ddof=2)
print(f'H0: потеря веса ~ N({mu_hat:.2f}; {sigma_hat:.2f}²)')
print('Наблюдаемые частоты:', observed)
print('Ожидаемые частоты:  ', expected.round(1))
print(f'χ² = {chi2_gof:.3f}, df = {k - 3}, p = {p4a:.4f}')
print(f'Вывод: {decision(p4a)}')
results.append(('χ² согласие (нормальность)', p4a))

fig, ax = plt.subplots(figsize=(10, 4))
pos = np.arange(k)
ax.bar(pos - 0.2, observed, width=0.4, label='Наблюдаемые', color='steelblue')
ax.bar(pos + 0.2, expected, width=0.4, label='Ожидаемые (нормальное)', color='salmon')
ax.set_xticks(pos)
ax.set_xticklabels([f'≤{e:.1f}' if np.isfinite(e) else f'>{edges[-2]:.1f}' for e in edges[1:]])
ax.set_xlabel('Интервал потери веса, кг'); ax.set_ylabel('Частота')
ax.set_title('Хи-квадрат: наблюдаемые и ожидаемые частоты'); ax.legend()
plt.tight_layout(); plt.show()
```

**Интерпретация.** χ² = 2.32, p = 0.80 — нет оснований отвергнуть гипотезу о нормальном распределении потери веса. Наблюдаемые частоты во всех 8 интервалах близки к ожидаемым 9.5. Вывод согласуется с тестами Шапиро–Уилка и Д'Агостино. При этом χ²-критерий при n = 76 обладает невысокой мощностью (данные огрубляются группировкой), поэтому основным аргументом за нормальность остаются Q-Q график и тест Шапиро–Уилка.

**(б) Независимость категориальных признаков.**

1. **Пол × диета** — проверка качества рандомизации: если участников распределяли по диетам случайно,
   доля мужчин и женщин в группах должна быть одинаковой. $H_0$: пол и диета независимы.
2. **Диета × достижение клинически значимого результата.** В медицине значимым считается снижение веса
   на **≥ 5%** от исходного. Связан ли тип диеты с тем, достиг ли участник этого порога?
   $H_0$: доля достигших порога одинакова для всех диет.

Размер эффекта — коэффициент V Крамера.

```python
df['success_5pct'] = np.where(df['loss_pct'] >= 5, '≥ 5%', '< 5%')

def chi2_independence(col1, col2):
    table = pd.crosstab(df[col1], df[col2])
    chi2, p, dof, exp = stats.chi2_contingency(table)
    v = stats.contingency.association(table.values, method='cramer')
    print(f'=== {col1} × {col2} ===')
    print(table, '\n')
    print('Ожидаемые частоты:')
    print(pd.DataFrame(exp, index=table.index, columns=table.columns).round(1), '\n')
    print(f'Минимальная ожидаемая частота = {exp.min():.1f} (условие E ≥ 5 {"выполнено" if exp.min() >= 5 else "НЕ выполнено"})')
    print(f'χ² = {chi2:.3f}, df = {dof}, p = {p:.4f}, V Крамера = {v:.3f}')
    print(f'Вывод: {decision(p)}\n')
    return p, table

p4b, tab_gender = chi2_independence('gender', 'diet')
p4c, tab_success = chi2_independence('diet', 'success_5pct')
results += [('χ² независимость (пол × диета)', p4b), ('χ² независимость (диета × успех ≥5%)', p4c)]
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
(tab_gender.div(tab_gender.sum(axis=0), axis=1).T * 100).plot(kind='bar', stacked=True, ax=axes[0],
                                                               colormap='Set3', rot=0)
axes[0].set_title('Состав групп по полу'); axes[0].set_ylabel('%'); axes[0].set_xlabel('')
axes[0].legend(title='', loc='lower right')
(tab_success.div(tab_success.sum(axis=1), axis=0) * 100).plot(kind='bar', stacked=True, ax=axes[1],
                                                               color=['lightgray', 'mediumseagreen'], rot=0)
axes[1].set_title('Доля участников, сбросивших ≥ 5% веса'); axes[1].set_ylabel('%'); axes[1].set_xlabel('')
axes[1].legend(title='', loc='lower right')
plt.tight_layout(); plt.show()
```

**Интерпретация.**

* **Пол × диета:** p = 0.98, V = 0.02 — связи нет, доля мужчин во всех группах ≈ 42–44%. Рандомизация проведена качественно: группы сопоставимы по полу, и различия в потере веса между диетами нельзя объяснить разным составом групп.
* **Диета × снижение веса ≥ 5%:** p = 0.013, V = 0.34 (связь средней силы). Клинически значимого результата достигли **70%** участников диеты 3, против **48%** на диете 2 и **29%** на диете 1. Этот результат согласуется с t-тестом и ANOVA, но см. следующий пункт: после поправки на множественность он оказывается на границе значимости.

### 5.5. Поправка на множественные сравнения

На одних данных проведено несколько тестов, поэтому вероятность хотя бы одной ложноположительной
находки выше 5%. Применяем поправку Бонферрони и более мощный метод Холма.

```python
names = [r[0] for r in results]
pvals = np.array([r[1] for r in results])
m = len(pvals)

p_bonf = np.minimum(pvals * m, 1)
order = np.argsort(pvals); p_holm = np.empty(m); running = 0
for rank, i in enumerate(order):
    running = max(running, (m - rank) * pvals[i]); p_holm[i] = min(running, 1)

mc = pd.DataFrame({'Тест': names, 'p': pvals, 'p (Бонферрони)': p_bonf, 'p (Холм)': p_holm})
mc['Значим без поправки'] = mc['p'] < ALPHA
mc['Значим (Холм)'] = mc['p (Холм)'] < ALPHA
print(f'Число тестов m = {m}; без поправки P(≥ 1 ложного отвержения) ≤ 1 − 0.95^{m} = {1 - 0.95 ** m:.3f}\n')
print(mc.to_string(index=False, float_format=lambda v: f'{v:.3g}'))
```

**Интерпретация.** Проведено 7 тестов, без поправки вероятность хотя бы одного ложноположительного вывода могла достигать ~30%.

* После поправки Холма значимыми остаются: снижение веса до/после (p ≈ 3·10⁻²¹), превышение потери в 3 кг (p = 0.009) и превосходство диеты 3 над диетой 1 (p = 0.033).
* Связь диеты с достижением порога 5% (p = 0.013 без поправки) после поправки Холма даёт p = 0.052, а по Бонферрони — 0.091, т.е. **перестаёт быть значимой** на уровне 0.05. Этот вывод нужно считать предварительным. Впрочем, он по сути дублирует t-тест «диета 3 vs 1» (это тот же эффект, только переведённый в бинарный признак с потерей информации), поэтому основным доказательством преимущества диеты 3 остаются t-тест и ANOVA.
* Незначимые результаты (пол, нормальность, баланс рандомизации) остаются незначимыми.

### Выводы

**1. Что показывают доверительные интервалы и p-значения в контексте задачи.**

* За 6 недель участники в среднем потеряли **3.95 кг** (95% ДИ: 3.37–4.52 кг; медиана 3.7 кг, ДИ 3.2–4.3 кг). Снижение веса статистически высоко значимо (парный t-тест, p ≈ 10⁻²²) и в среднем превышает «безопасный» темп 3 кг за 6 недель (p = 0.0015), хотя и ненамного.
* **Диета 3 эффективнее** диет 1 и 2: в среднем на ~1.85 кг больше (95% ДИ разности с диетой 1: 0.54–3.16 кг, d ≈ 0.8), что подтверждено t-тестом, ANOVA, непараметрическими критериями и бутстрепом, и сохраняется после поправки Холма. Диеты 1 и 2 между собой не различаются.
* Эффективность диет **не зависит от пола** (разность 0.12 кг, ДИ содержит 0).
* Рандомизация корректна: группы сбалансированы по полу (χ², p = 0.98).
* Доля участников, сбросивших ≥ 5% веса, выше на диете 3 (70% против 29–48%), но после поправки на множественность этот результат на границе значимости.

**2. Влияет ли выбор метода (параметрический vs бутстреп) на выводы?** На **выводы — нет**: все интервалы для среднего и разностей средних согласованы, ни один вывод о значимости не меняется. Но на **ширину интервалов** — да: бутстреп систематически даёт интервалы на 2–3% уже по всей выборке и на ~7% уже в малых группах (n ≈ 25), т.е. немного переоценивает точность. Поскольку данные нормальны, параметрические t-интервалы здесь точнее и предпочтительнее. Сильнее всего методы расходятся для дисперсии — статистики с асимметричным распределением; там лучше всех из бутстреп-методов работает BCa. Бутстреп незаменим для медианы, для которой простой параметрической формулы нет.

**3. Какой тест наиболее адекватен данным.** **Парный t-тест** — данные изначально собраны по схеме «до/после» на одних и тех же людях, разности нормальны (Шапиро p = 0.79), вывод подтверждён критерием Уилкоксона; использование независимого теста здесь было бы ошибкой. Для главного вопроса исследования («какая диета лучше») наиболее адекватен **дисперсионный анализ с попарными t-тестами и поправкой Холма**: групп три, дисперсии равны, распределения в группах нормальны. Одиночный двухвыборочный t-тест корректен, но выбор пары «диета 3 vs 1» сделан после просмотра данных, поэтому его p-значение нужно корректировать.

**Ограничения:** малый объём выборки (по ~25 человек на диету); нет контрольной группы без диеты; 2 участника исключены как вероятно выбывшие; длительность наблюдения — всего 6 недель, о долгосрочном эффекте данные ничего не говорят.
